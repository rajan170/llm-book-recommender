```python
"""
This cell contains the implementation of the function `calculate_emotion_scores` which calculates the maximum emotion scores for each emotion label from the given predictions.
The `emotion_labels` list contains the possible emotion labels: "anger", "disgust", "fear", "joy", "sadness", "surprise", and "neutral".
The `isbn` list is initialized as an empty list, which can be used to store ISBN numbers of books if needed.
The `emotion_scores` dictionary is initialized with keys as emotion labels and values as empty lists, which can be used to store emotion scores for each label.
The `calculate_emotion_scores` function takes a list of predictions as input. Each prediction is a list of dictionaries, where each dictionary contains an emotion label and its corresponding score.
The function processes the predictions to calculate the maximum score for each emotion label. It returns a dictionary with emotion labels as keys and their corresponding maximum scores as values.
"""

import numpy as np

emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise", "neutral"]
isbn = []
emotion_scores = {label: [] for label in emotion_labels}

def calculate_max_emotion_scores(predictions):
    per_emotion_scores = {label: [] for label in emotion_labels}
    for prediction in predictions:
        sorted_predictions = sorted(prediction, key=lambda x: x["label"])
        for index, label in enumerate(emotion_labels):
            per_emotion_scores[label].append(sorted_predictions[index]["score"])
    return {label: np.max(scores) for label, scores in per_emotion_scores.items()}
```

```python
for i in range(10):
    isbn.append(books["isbn13"][i])
    sentences = books["description"][i].split(".")
    predictions = classifier(sentences)

    max_scores = calculate_max_emotion_scores(predictions)
    
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])
```

```python
# The following code block initializes the emotion labels, ISBN list, and emotion scores dictionary.
# It then iterates over the range of the length of the books DataFrame, appending the ISBN numbers to the ISBN list.
# For each book description, it splits the description into sentences and uses the classifier to predict the emotions for each sentence.
# The maximum emotion scores are calculated using the calculate_max_emotion_scores function.
# These maximum scores are then appended to the corresponding lists in the emotion_scores dictionary.

from tqdm import tqdm

 # Initialize emotion labels, ISBN list, and emotion scores dictionary
emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise", "neutral"]
isbn = []
emotion_scores = {label: [] for label in emotion_labels}

# Iterate over the range of the length of the books DataFrame
for i in tqdm(range(len(books))):
    # Append the ISBN number to the ISBN list
    isbn.append(books["isbn13"][i])
    
    # Split the book description into sentences
    sentences = books["description"][i].split(".")
    
    # Predict the emotions for each sentence using the classifier
    predictions = classifier(sentences)
    
    # Calculate the maximum emotion scores
    max_scores = calculate_max_emotion_scores(predictions)
    
    # Append the maximum scores to the corresponding lists in the emotion_scores dictionary
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])
```

```python
emotions_df = pd.DataFrame(emotion_scores)
emotions_df["isbn13"] = isbn
```

```python
emotions_df
```

```python
books = pd.merge(books, emotions_df, on = "isbn13")
```
